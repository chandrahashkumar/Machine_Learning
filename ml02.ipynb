```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    'Name': ['Ram', 'Shyam','Rohit', 'Kal', 'AAKASH'],
    'Age':[19,20,22,25,23],
    'Salary': [45000,46000,70000,56000,456000]
})

df.head()
```

# ***Plot***

```python
plt.scatter(x=df['Age'], y=df['Salary'], color='blue')
plt.title('Age vs Salary')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.show()
```
